```python
import numpy as np
import matplotlib.pyplot as plt

from obspy import UTCDateTime
from obspy.clients.fdsn import Client

client = Client("IRISDMC")

stations = ["DUCK1", "DUCK2", "DUCK3"]

start = UTCDateTime("2026-09-18T00:00:00")
end = UTCDateTime("2026-09-18T04:00:00")
```

```python
data = {}

for station in stations:

    st = client.get_waveforms(
        network="UO",
        station=station,
        location="*",
        channel="HNZ",
        starttime=start,
        endtime=end
    )

    data[station] = st

    print(f"{station}:")
    print(st)
```
